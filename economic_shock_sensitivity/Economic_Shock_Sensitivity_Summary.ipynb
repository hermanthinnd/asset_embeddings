# Do Holdings Embeddings Reveal Economic-Shock Sensitivity?

## Presentation summary

This extension asks whether the geometry learned from institutional holdings contains an economically interpretable signal: **which firms are relatively helped or hurt by unexpected monetary tightening?** It uses 2005–2024 holdings and return data, 172 high-frequency FOMC surprises, 60 quarterly prediction problems from 2010Q1 through 2024Q4, and a permanently held-out 2021–2024 test period.

**Main conclusion.** Holdings embeddings contain incremental ex-ante information about firm rate sensitivity beyond six standard characteristics. The result is strongest and most stable for RS-L-Min. The evidence supports *latent exposure measurement*, not a claim that embeddings forecast large abnormal returns or that funds systematically trade on the recovered direction after each surprise.

## 1. Research design

The experiment proceeds in a strict temporal sequence:

1. Measure unexpected policy news using the San Francisco Fed monetary-policy-surprise data. Positive STMT values mean an unexpected tightening or increase in the one-year Treasury yield in the 30-minute FOMC statement window.
2. Build PERMCO-level event returns from CRSP, using one day before through one day after each announcement and lagged market equity to aggregate multiple securities.
3. At every quarter, estimate each firm's historical rate beta using only the preceding five years of FOMC events. The primary estimand is the raw beta, clipped at training-sample 1st and 99th percentiles inside each prediction problem.
4. Fit Ridge models on fixed firm splits. Hyperparameters are selected in the development period, checked in validation, and then frozen before the 2021–2024 final test.
5. Compare characteristics, RS-Binary, RS-L-Min, and embeddings plus characteristics.
6. Challenge the incremental result with paired quarterly inference and 100 shuffled-embedding placebos.
7. Ask which mutual-fund behavior creates the most informative embedding geometry.
8. Finally, test whether funds actively rebalance along the pre-existing rate-sensitivity direction after shocks.

Two timing conventions are reported. **q−1** treats the prior quarter's holdings as economically predetermined. **q−2** is the conservative public-information design, allowing for filing delay. The close agreement between them is an important anti-leakage check.

## 2. Data and measurement audit

- 172 FOMC announcement events and 515 required CRSP trading dates.
- 3,869,729 raw security-date rows; 1,478 exact duplicates removed and no conflicting duplicates.
- 1,022,225 firm-event observations across 14,298 firms.
- Median event cross-section: 5,722 firms.
- Median embedding-universe return coverage from 2010 onward: 99.36%.
- Historical target requires at least 20 events in the preceding five years.

The target itself has genuine forward content. In the full sample, the historical STMT beta predicts later event returns with an interaction slope of **0.109**, two-way-clustered t = **2.89**. The development and validation estimates are positive (t = 2.95 and 2.13); the untouched final-period estimate remains positive but is imprecise (0.057, t = 0.98). A 200-draw shuffled-beta test gives p = **0.00498**. ME-shock and alternative-return-window results show that measurement choice matters, so the economic claim is deliberately narrow.

## 3. Primary ex-ante result



Under conservative q−2 timing, characteristics explain **5.73%** of final-test variance. RS-Binary alone explains **4.29%** and RS-L-Min alone **5.40%**. Combining the representations raises performance to **7.25%** for Binary plus characteristics and **8.04%** for L-Min plus characteristics. The q−1 conclusions are almost identical: 5.87%, 3.81%, 5.25%, 7.07%, and 8.14%, respectively.

This establishes complementarity rather than replacement: embeddings capture structure absent from conventional characteristics, while characteristics retain useful information absent from embeddings.

## 4. Robustness of the incremental contribution



For RS-L-Min plus characteristics, the paired quarter-by-quarter gain over characteristics is **2.34 percentage points** under q−1 timing and **2.36 points** under q−2 timing. Every one of the 16 final-test quarters is positive. Newey–West t-statistics are **8.58** and **9.52**, and moving-block 95% intervals are [1.92, 3.05] and [1.84, 2.92] percentage points.

In contrast, shuffled embeddings reduce performance on average. The observed gains exceed every one of the 100 shuffled draws under both timings (finite-sample permutation p = **0.0099**). RS-Binary also adds statistically reliable information, but its final-period gain is smaller: 1.31 points at q−1 and 1.56 points at q−2.

These checks address the central alternative explanation: the gain is not merely caused by adding ten arbitrary predictors or by fitting the embedding on the same firm universe.

## 5. Which investors make the signal visible?



The behavioral comparison is not a simple active-versus-passive horse race. It asks which subset of mutual-fund holdings creates a geometry most aligned with rate sensitivity. Under q−1 timing, stable stock-picker RS-Binary achieves **11.17%** final-test pooled R²—the strongest behavioral result. Tactical-active Binary reaches 9.06%; passive-like L-Min reaches 9.01%; and benchmark-rebalancer Binary reaches 8.55%. Results remain similar with q−2 inputs.

Interpretation: persistent, selective co-ownership appears especially informative for the binary representation, while passive-like holdings can still carry useful broad-market structure. Active management is therefore not uniformly superior; the interaction between investor behavior and embedding construction matters.

## 6. Post-shock trading is a separate question

The final test constructs a passive counterfactual for every S12 fund. Prior-quarter weights are drifted by realized stock returns; actual ending weights minus those passive weights measure active trading. That trade vector is projected onto rate betas estimated before the quarter. Quarter-end holdings are treated as outcomes disclosed in q+1, never as predictors of the shock that already occurred.

Across **297,941 fund-quarter observations** and all 60 quarters, the average response to the STMT shock is positive but insignificant: slope **0.00413**, Newey–West t = **1.04**, p = **0.299**. In the final 2021–2024 period it is effectively zero: slope −0.00118, t = −0.26. No behavioral subgroup has a statistically reliable response. All coverage and weight-balance integrity gates pass.

This null result sharpens the conclusion. Holdings contain a stable cross-sectional map of economic exposure, but mutual funds do not appear to systematically rotate their portfolios along that map immediately after monetary-policy surprises.

## 7. What the project establishes—and what it does not

### Supported

- Holdings geometry recovers an economically interpretable firm characteristic out of sample.
- RS-L-Min and conventional characteristics are complementary.
- The incremental contribution is stable across all final quarters, both timing conventions, and shuffled-feature placebos.
- The informativeness of embeddings varies meaningfully with investor behavior.

### Not established

- The final-period direct return interaction is not statistically significant.
- The exercise does not establish a profitable tradable strategy after transaction costs.
- It does not show that investors consciously know or trade the latent rate direction.
- The post-shock holdings test is descriptive and disclosed with delay.

The defensible final claim is therefore: **institutional-holdings embeddings encode persistent cross-sectional information about firms' sensitivity to monetary-policy shocks, above standard characteristics, even though direct final-period return and fund-rebalancing evidence is weaker.**
